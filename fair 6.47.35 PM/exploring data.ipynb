```python
import pandas as pd
from pandas import json_normalize
import json
```

```python
def read_json_fair():
    try:
        with open("fair_results.json") as f:
            data = json.load(f)
        
        json_list = json_normalize(data)
        list_dicts = json_list[0].tolist()
        return pd.DataFrame(list_dicts)
    except Exception as e:
        print(f'Not able to read the fair results file: {e}')

def group_list(df):
    return df['group_name'].unique()

def dataset_type(df):
    return df['dataset_type'].unique()
```

```python
df = read_json_fair()
```
